```python
import meteostat as ms
import pandas as pd

# Date range
START = pd.Timestamp("2024-07-25")
END = pd.Timestamp("2026-04-18")

# Meteostat station
station_id = "41194"

# Fetch daily weather
ts = ms.daily(station_id, START, END)
weather = ts.fetch().reset_index()

# Rename columns to clearer names
weather = weather.rename(columns={
    "time": "date",
    "temp": "temp_meteostat",
    "tmin": "temp_min_meteostat",
    "tmax": "temp_max_meteostat",
    "rhum": "humidity_meteostat",
    "prcp": "precipitation_meteostat",
    "snwd": "snow_depth_meteostat",
    "wspd": "wind_speed_meteostat",
    "wpgt": "wind_gust_meteostat",
    "pres": "pressure_meteostat",
    "tsun": "sunshine_meteostat",
    "cldc": "cloud_cover_meteostat"
})

# Convert date to plain date
weather["date"] = pd.to_datetime(weather["date"]).dt.date

# Sort by date
weather = weather.sort_values("date").reset_index(drop=True)

# Save to file
weather.to_csv("dubai_meteostat_daily.csv", index=False)

print("Saved: dubai_meteostat_daily.csv")
print(weather.head())
print(weather.columns.tolist())
```
